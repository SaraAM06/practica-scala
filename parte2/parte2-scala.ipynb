## Ejercicio 1 — Variables, tipos e inferencia
En este ejercicio vamos a representar los datos de una partida de cartas utilizando dos enfoques: uno con declaración explícita de tipos (Versión A) y otro dejando que Scala los infiera automáticamente (Versión B).

```scala
val nombre: String = "Sara"
var partidas: Int = 3
var puntuacion: Int = 20
var porcentaje_victorias: Double = 34.23
var activo: Boolean = true

```

```scala
val nombre  = "Sara"
val partidas = 3
var puntuacion = 20
var porcentaje_victorias = 34.23
var activo = true
```

### Conclusiones y Documentación del Ejercicio 1

**Qué tipos ha inferido Scala:** Scala ha inferido automáticamente `String` para el nombre, `Int` para las partidas y la puntuación, `Double` para el porcentaje de victorias, y `Boolean` para el indicador de actividad.

**Qué diferencia observas entre ambas versiones:** La Versión A emplea tipado explícito, declarando el tipo de dato directamente. La Versión B se apoya en el sistema de inferencia estática de Scala, gracias al cual el compilador deduce el tipo a partir del dato introducido.

**Qué variables deberían ser val y cuáles podrían ser var:** Se ha utilizado `val` para el nombre del jugador ya que su identidad no varía durante la partida. Por el contrario, se han empleado `var` en variables como la puntuación o el porcentaje para reflejar que son valores susceptibles de cambiar o actualizarse dinámicamente ronda a ronda.

# Ejercicio 5 — Comparación de dos manos
Comparar dos manos distintas y ver cual es la mayor de ellas y, por tanto, gana o va ganando.

```scala
def maxHand(handA: Int, handB: Int): Int = {
    if (handA > handB) handA
    else if (handB > handA) handB
    else 0
}

maxHand(17,19)
maxHand(20,18)
maxHand(21,21)
```

### Conclusiones y Documentación del Ejercicio 5

# Ejercicio 6 — Decidir el ganador de una partida
Estudiar quién es el ganador, repasando si se han pasado y, de no haberse pasado, el que tiene el número mayor.

```scala
def bust(hand: Int): Boolean = {
    hand > 21
}

def maxHand(handA: Int, handB: Int): Int = {
    if (handA > handB) handA
    else if (handB > handA) handB
    else 0
}

def ganador(handA: Int, handB: Int): Int = {
    if (bust(handA) && bust(handB)) 0
    else if (bust(handA)) handB
    else if (bust(handB)) handA
    else maxHand(handA, handB)
}

ganador(26,20)
ganador(18,22)
ganador(24,25)
ganador(17,19)
ganador(21,20)
```

### Conclusiones y Documentación del Ejercicio 6
En el primer ejemplo se cumple el primer `else if` pues handA supera 21.
En el segundo ejemplo se cumple el segundo `else if` pues handB supera 21.
En el tercer ejemplo se cumple el primer `if` pues ambos se pasan de 21.
En el cuarto ejemplo se cumple el último `else` pues ninguno se pasa de 21 y llama a la función maxHand.
En el quinto ejemplo se cumple el último `else` pues ninguno se pasa de 21 y llama a la función maxHand.

# Ejercicio 7 — Arrays y mutabilidad
Modificar arrays y estudiar las diferencias entre reasignación y modificación

```scala
val jugadores = Array("Alex", "Chen", "Marta")
println(jugadores)
jugadores(0) = "Sindhu"
println(jugadores)
jugadores(0) = 500
```

### Conclusiones y Documentación del Ejercicio 7
**Modificación con `val`:** `val` solo impide cambiar la referencia del array, pero el contenido interno del objeto sigue siendo mutable.

**Tipado estricto:** `Array[String]` exige estrictamente datos de tipo `String`. El valor `500` es un `Int`, por lo que el compilador bloquea la asignación por incompatibilidad de tipos.

**Reasignación vs Modificación:** Reasignar cambia el objeto al que apunta la variable (exige `var`), mientras que modificar un elemento solo altera el contenido interno del array existente.

# Ejercicio 9 — Recorrer un Array con while


```scala
val manos = Array(17, 24, 21, 19, 26)

def bust(hand: Int): Boolean = {
    hand > 21
}

var i = 0
while (i < manos.length) {
    if(i != manos.length - 1) {
        println(manos(i) + " " + bust(manos(i)))
    }
    i = i + 1
}
```

### Conclusiones y Documentación del Ejercicio 9
**Por qué el contador debe ser mutable y qué ocurriría si no incrementaras** Porque debe modificarse cada ronda, sino fuese así, el bucle sería infinito.

**Qué elementos del ejercicio corresponden al estilo imperativo** 
    La variable mutable var i = 0 (el contador de posición). 
    El bucle while.
    La instrucción i = i + 1 para cambiar el valor del contador paso a paso.

# Ejercicio 10 — Listas e inmutabilidad
Crear nuevas listas mediante ´::´ y uso de funciones para obtener la longitud de una lista y devolver una versión invertida de ésta


```scala
val jugadores = List("Alex", "Chen", "Marta")
println(jugadores)

val jugadores_dos = "Sindhu" :: jugadores

println("Lista original: " + jugadores)
println("Lista modificada: " + jugadores_dos)

println("Lista original longitud: " + jugadores.length)
println("Lista modificada longitud: " + jugadores_dos.length)

println("Lista original invertida: " + jugadores.reverse)
println("Lista modificada invertida: " + jugadores_dos.reverse)
```

### Conclusiones y Documentación del Ejercicio 9
En Scala, la operación con :: produce una lista nueva en lugar de modificar la original porque las listas son inmutables por defecto.

# Ejercicio 13 — foreach y funciones como valores
Método tradicional y funcional de recorrer un mismo array para comprobar que valores de la lista se han pasado de 21

```scala
val manos = Array(17, 24, 21, 26, 18)

def bust(hand: Int): Unit = {
    println(hand > 21)
}
println("RECORRER EL ARRAY CON foreach")
manos.foreach(bust)

println()
println("RECORRER EL ARRAY CON while")
var i = 0
while (i < manos.length) {
   bust(manos(i))
    i = i + 1
}
```

### Conclusiones y Documentación del Ejercicio 13
**¿Cuál necesita un contador?** El bucle while

**¿Cuál necesita una variable var para recorrer la colección?** El bucle while, el propio contador de la pregunta anterior.

**Cuál se aproxima más al estilo funcional presentado en el material del curso** El método foreach pues reduce mucho la cantidad de código

# Ejercicio 15 — Programa integrado: torneo de Twenty-One
Poner a prueba los conocimientos adquiridos

```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)

def bust(hand: Int): Boolean = {
    hand > 21
}

def mejorMano(puntuaciones: Array[Int]): Int = {
    var i = 0
    var mayor_valor = 0
    while (i < puntuaciones.length) {
        if (!bust(puntuaciones(i)) && puntuaciones(i) > mayor_valor) {
            mayor_valor = puntuaciones(i)
        }
        i = i + 1
    }
    mayor_valor
}

println("¿Qué manos se han pasado?")
manos.foreach(hand => println(bust(hand)))

println("¿Cuál es la mejor mano válida?")
println(mejorMano(manos))
println()

val manosRonda2 = Array(22, 19, 20, 21)

println("¿Qué jugadores continúan con puntuación válida?")
var j = 0
while (j < manosRonda2.length) {
    if (!bust(manosRonda2(j))) {
        println(jugadores(j) + " sigue en partida")
    }
    j = j + 1
}

println("¿Cuál es la mejor mano válida?")
println(mejorMano(manosRonda2))
println()
```

### Conclusiones y Documentación del Ejercicio 15
**Partes mutables:** Las variables declaradas con `var` (los contadores `i`, `j` y el acumulador `mayor_valor`), cuyos valores cambian durante la ejecución de los bucles.

**Partes inmutables:** Las colecciones declaradas con `val` (`jugadores`, `manos` y `manosRonda2`), cuyas referencias y datos permanecen intactos tras su creación.

**Funciones:** La función `def bust` para verificar el límite de 21 y la función encapsulada `def mejorMano` para calcular la puntuación máxima válida de forma reutilizable.

**Estructuras de control:** El método `foreach` y los bucles `while` para la iteración, combinados con bloques condicionales `if` y operadores lógicos (`&&`, `!`) para filtrar las manos.

**Estilo funcional:** El uso de métodos de orden superior como `foreach` y la división del programa en funciones especializadas y modulares.

**Estilo imperativo:** El uso de bucles `while` tradicionales basados en contadores incrementales y la mutación manual de variables de estado (`mayor_valor`).
